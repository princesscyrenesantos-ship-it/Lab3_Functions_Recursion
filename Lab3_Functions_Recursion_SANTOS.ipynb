```python
# [LAB 3 - TASK 1.1] STUDENT IDENTIFICATION
LAST_NAME = "SANTOS"
STUDENT_ID = "TUPM-26-1866"
SEED_DIGIT = int(STUDENT_ID[-1])

print("Student Last Name:", LAST_NAME)
print("Student ID:", STUDENT_ID)
print("Seed Digit:", SEED_DIGIT)
```

```python
# [LAB 3 - TASK 1.2] FUNCTION DEFINITIONS & CALLS
def greet_student(name, student_id):
    return f"Welcome to Lab 3, {name} ({student_id})!"

def calculate_base_score(seed):
    return (seed * 10) + 50

welcome_msg = greet_student(LAST_NAME, STUDENT_ID)
base_score = calculate_base_score(SEED_DIGIT)

print(welcome_msg)
print("Base Score calculated:", base_score)
```

```python
# [LAB 3 - TASK 1.3] VARIABLE SCOPE & DEFAULTS
global_bonus = 5

def compute_total_score(score, bonus=global_bonus):
    local_multiplier = 1.1
    return (score + bonus) * local_multiplier

total_score = compute_total_score(base_score)
print(f"Total Score (with default bonus): {total_score:.2f}")
```

```python
# [LAB 3 - TASK 2.1] FUNCTION ENCAPSULATION & ARGS
def process_student_data(name, student_id, *scores):
    avg_score = sum(scores) / len(scores) if scores else 0
    return {
        "Student": name,
        "ID": student_id,
        "Average": round(avg_score, 2),
        "Passed": avg_score >= 75
    }

result = process_student_data(LAST_NAME, STUDENT_ID, 85, 90, 78, 88 + SEED_DIGIT)
print("Task 2.1 Result:", result)
```

```python
# [LAB 3 - TASK 2.2] MODULE IMPORTATION
import grades

scores_list = [88, 92, 79, 85 + SEED_DIGIT]
avg = grades.get_average(scores_list)
final_grade = grades.calculate_grade(avg)

print(f"Calculated Average: {avg:.2f}")
print(f"Final Letter Grade: {final_grade}")
```

```python
# [LAB 3 - TASK 2.3] FUNCTIONAL UTILITIES (LAMBDA & MAP)
raw_scores = [70, 82, 91, 65, 88]
curved_scores = list(map(lambda x: min(100, x + SEED_DIGIT + 2), raw_scores))

print("Original Scores:", raw_scores)
print("Curved Scores:", curved_scores)
```

```python
# [LAB 3 - TASK 3.1] RECURSIVE FACTORIAL
def recursive_factorial(n):
    if n <= 1:
        return 1
    return n * recursive_factorial(n - 1)

fact_input = SEED_DIGIT if SEED_DIGIT > 0 else 5
fact_result = recursive_factorial(fact_input)

print(f"Factorial of {fact_input}! =", fact_result)
```

```python
# [LAB 3 - TASK 3.2] RECURSIVE FIBONACCI
def recursive_fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return recursive_fibonacci(n - 1) + recursive_fibonacci(n - 2)

fib_input = SEED_DIGIT + 2
fib_result = recursive_fibonacci(fib_input)

print(f"Fibonacci term at position {fib_input}:", fib_result)
```

```python
# [LAB 3 - TASK 3.3] RECURSIVE DIGIT SUM
def recursive_sum_digits(n):
    if n < 10:
        return n
    return (n % 10) + recursive_sum_digits(n // 10)

clean_id_num = int("".join(filter(str.isdigit, STUDENT_ID)))
digit_sum_result = recursive_sum_digits(clean_id_num)

print(f"Recursive Sum of Digits for {STUDENT_ID}:", digit_sum_result)
```
